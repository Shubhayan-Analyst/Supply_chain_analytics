```python

raw_dates = df['shipping_date']
dash_count = raw_dates.str.contains('-').sum()
slash_count = raw_dates.str.contains('/').sum()
print(dash_count, slash_count)
```

```python
dash_mask = raw_dates.str.contains('-')

parsed_dash = pd.to_datetime(raw_dates[dash_mask], format='%d-%m-%Y %H:%M', errors='coerce')
parsed_slash = pd.to_datetime(raw_dates[~dash_mask], format='%m/%d/%Y %H:%M', errors='coerce')

df['shipping_date'] = pd.concat([parsed_dash, parsed_slash]).sort_index()
```
